```python
# BRFSS Diabetes Data Audit

## 1. Project Setup

from pathlib import Path
import pandas as pd
import numpy as np

RANDOM_STATE = 42

DATA_PATH = Path("data/diabetes_binary_5050split_health_indicators_BRFSS2015.csv")
print("Random state:", RANDOM_STATE)
print("Dataset path:", DATA_PATH.resolve())

```

```python
## 2. Load Dataset
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
## 3. Dataset Dimensions
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
df.tail()
```

```python
## 4. Target Distribution
target = "Diabetes_binary"

target_counts = df[target].value_counts(dropna=False).sort_index()
target_percentages = df[target].value_counts(normalize=True, dropna=False).sort_index() * 100

target_summary = pd.DataFrame({
    "count": target_counts,
    "percentage": target_percentages
})

target_summary
```

```python
## 5. Missing Values
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": df.isna().mean() * 100
})

missing_summary

```

```python
print("Total missing values:", df.isna().sum().sum())
```

```python
## 6. Data Types
dtype_summary = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "non_null": df.notna().sum(),
    "unique_values": df.nunique(dropna=False)
})

dtype_summary
```

```python
## 7. Unique Values and Coding
for column in df.columns:
    values = sorted(df[column].dropna().unique().tolist())
    print(f"{column}: {values}")
```

```python
## 8. Potential Duplicate Profiles
duplicate_count = df.duplicated().sum()
duplicate_percentage = duplicate_count / len(df) * 100

print("Fully duplicated rows:", duplicate_count)
print(f"Percentage of rows duplicated: {duplicate_percentage:.2f}%")
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Rows participating in duplicate groups:", len(duplicate_rows))
duplicate_rows.head(20)
```

```python
## 9. Data-Audit Summary
audit_summary = {
    "rows": df.shape[0],
    "columns": df.shape[1],
    "predictors": df.shape[1] - 1,
    "target": target,
    "class_0": int(target_counts.get(0, 0)),
    "class_1": int(target_counts.get(1, 0)),
    "total_missing": int(df.isna().sum().sum()),
    "fully_duplicated_rows": int(duplicate_count),
    "duplicate_percentage": round(duplicate_percentage, 2)
}

audit_summary
```

```python
audit_summary = {
    "rows": int(df.shape[0]),
    "columns": int(df.shape[1]),
    "predictors": int(df.shape[1] - 1),
    "target": target,
    "class_0": int(target_counts.get(0, 0)),
    "class_1": int(target_counts.get(1, 0)),
    "total_missing": int(df.isna().sum().sum()),
    "fully_duplicated_rows": int(duplicate_count),
    "rows_in_duplicate_groups": int(len(duplicate_rows)),
    "duplicate_percentage": float(round(duplicate_percentage, 2))
}

audit_summary
```
